# Categorical Data

```python
#  so far we have worked with numbers float string boolean
# Categorical data in the form of group (or descrete categories)
#  groups could be labled or textual 
```

```python
df.head()
#  In the below given data e.g age anf name cannot be a catagory
#  while sex, embarked ,pclass can be catagories
```

```python
print(df['Sex'].dtype)
```

```python
# Convert a column to categorical
df['Sex_C'] = df['Sex'].astype('category')
```

```python
df[['Sex', 'Sex_C']].head()
```

```python
print(df['Sex'].dtype)
print(df['Sex_C'].dtype)
```

```python
#  We will be able to derive more things if the datatype is catagory
df['Sex_C_categorical_codes'] = df['Sex_C'].cat.codes
# df['Sex_C'].cat.codes will retrieve us the code 
```

```python
# Machine Learning systems doesnot understand strings. They understand 0,1, number or some big numbers
#  We cannot put male and female into machine learning 
#  that is why we are converting them to a catagorical data 
```

```python
df[['Sex', 'Sex_C', 'Sex_C_categorical_codes']].head()
```

```python
df['Sex_apply_category'] = df['Sex'].apply(lambda x : 0 if x == 'male' else 1)
# pythonic version
# we can set our own code 
```

```python
df[['Sex', 'Sex_C', 'Sex_C_categorical_codes', 'Sex_apply_category']].head()
```

```python
df.head()
```

```python
df['Pclass'].unique()
#  to find the unique values in this column
```

```python
df['Pclass_new_weird'] = df['Pclass'].apply(lambda x : 'adf' if x == 3 else ('wer' if x == 2 else 'tyu'))
```

```python
df[['Pclass', 'Pclass_new_weird']].head(10)
```

```python
#  If we don't have pclass and after analysis of Pclass_new_weird by reading articles about it  we found that 
# tyu > adf > wer => wer is the smallest 
#  but if all are string does this make sence for python or 3 > 1 like in numeric  make more sence 
#  numeric data has more sence becaue we don't have realtions in strings 
```

```python
print(df['Pclass'].dtype)
print(df['Pclass_new_weird'].dtype)
```

```python
df['Pclass_new_weird_ordered'] = pd.Categorical(
    df['Pclass_new_weird'],
    categories=['wer','adf', 'tyu'], # Custom order (ascending)
    ordered=True
)
```

```python
print(df['Pclass_new_weird_ordered'].dtype)
```

```python
df['Pclass_new_weird_ordered'].cat.categories
```

# Date Functionality in pandas

```python
df = pd.DataFrame({
    'event': ['Concert', 'Conference', 'Wedding', 'Reception', 'last_day'],
    'date': ['2025-01-01', '2025-03-15', '2025-07-20', '2025-07-21', '2025-07-31']
})
```

```python
df['date_pandas'] = pd.to_datetime(df['date'])
```

```python
print(df.dtypes[['event', 'date', 'date_pandas']])
```
